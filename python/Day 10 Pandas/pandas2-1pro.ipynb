```python
import pandas as pd
import numpy as np

sales = pd.DataFrame({
    "Product":  ["Laptop", "Phone", "Headphones", "Tablet",
                 "Phone", "Laptop", "Headphones", "Tablet"],
    "Category": ["Computers", "Mobiles", "Accessories", "Computers",
                 "Mobiles", "Computers", "Accessories", "Computers"],
    "Units":    [3, 10, 25, 6, 8, 2, 30, 4],
    "Price":    [55000, 20000, 1500, 25000, 22000, 58000, 1200, 27000],
})

# 1. Look at the data
print(sales.head())
print(sales.shape)
sales.info()

# 2. New column
sales["Revenue"] = sales["Units"] * sales["Price"]
print("Total revenue:", sales["Revenue"].sum())      # 988500

# 3. Best single sale
best = sales.loc[sales["Revenue"].idxmax()]
print("Best sale:\n", best)

# 4. Revenue per product, highest first
by_product = sales.groupby("Product")["Revenue"].sum().sort_values(ascending=False)
print(by_product)       # Phone 376000, Laptop 281000, Tablet 258000, Headphones 73500

# 5. Units sold per category
print(sales.groupby("Category")["Units"].sum())

# 6. Filter: big sales only
big = sales[sales["Revenue"] > 100000]
print(big[["Product", "Revenue"]])

# 7. Label each sale
sales["Size"] = np.where(sales["Revenue"] > 100000, "Big", "Small")
print(sales["Size"].value_counts())


sales.to_csv("sales_report.csv", index=False)
reloaded = pd.read_csv("sales_report.csv")
print(reloaded.equals(sales))
```
