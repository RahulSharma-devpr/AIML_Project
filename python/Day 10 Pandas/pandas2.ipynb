```python
#Series
import pandas as pd
marks = pd.Series([88, 72, 95, 60],
                  index=["Vikas", "Rahul", "Ayush", "Kabir"],
                  name="Maths")
print(marks)

print(marks["Vikas"])          
print(marks.iloc[1])           
print(marks[["Rahul", "Ayush"]])   
print(marks[marks > 70])       
print(marks.max())

print(marks.mean(), marks.max(), marks.idxmax())  
print(marks + 5)               
print(marks.sort_values(ascending=True))
```

```python
data = {
    "Name":  ["Ayush", "Rahul", "Meena", "Kabir", "Isha"],
    "Age":   [21, 22, 20, 23, 21],
    "City":  ["Delhi", "Mumbai", "Delhi", "Pune", "Mumbai"],
    "Marks": [88, 72, 95, 60, 81],
}
df = pd.DataFrame(data)
print(df)
```

```python
print(df.head(3))        # first 3 rows (default 5)
print(df.tail(2))        # last 2 rows
print(df.shape)          # (5, 4)   rows, columns
print(df.columns)        # Index(['Name', 'Age', 'City', 'Marks'], dtype='object')
print(df.dtypes)         # data type of each column
df.info()                # columns, types, non-null counts, memory
print(df.describe())     # count, mean, std, min, quartiles, max for numeric columns
```

```python
print(df["Name"])                  
print(df[["Name", "Marks"]])      
print(df.loc[0])                       
print(df.loc[0, "Name"])               
print(df.loc[1:3, ["Name", "Marks"]])  


print(df.iloc[0:2, 0:2])               
print(df.iloc[-1])                     
```

```python
print(df[df["Marks"] > 80])                              # Asha, Meena, Isha
print(df[(df["City"] == "Delhi") & (df["Marks"] > 90)])  # Meena
print(df[(df["Age"] < 21) | (df["Marks"] < 65)])         # Meena, Kabir
print(df[df["City"].isin(["Delhi", "Pune"])])            # Asha, Meena, Kabir
print(df[df["Name"].str.startswith("A")])                # Asha
```

```python
df["Grade"] = np.where(df["Marks"] >= 85, "A", "B")    # new column from a condition
df["Bonus"] = df["Marks"] + 5                          # new column from maths
df.loc[df["Name"] == "Kabir", "Marks"] = 65            # change one value

df = df.drop(columns=["Bonus"])                        # remove a column
df = df.rename(columns={"Marks": "Score"})             # rename a column
```

```python
print(df.sort_values("Score", ascending=False))
print(df.sort_values(["City", "Score"], ascending=[True, False]))

print(df["Score"].mean(), df["Score"].median(), df["Score"].max())
print(df["City"].value_counts())         # Delhi 2, Mumbai 2, Pune 1

# A first look at groupby: average score per city
print(df.groupby("City")["Score"].mean())    # Delhi 91.5, Mumbai 76.5, Pune 65.0
```

```python
messy = pd.DataFrame({"Name": ["A", "B", "C"],
                      "Score": [80, np.nan, 90]})

print(messy.isna())                # True where a value is missing
print(messy.isna().sum())          # missing count per column
print(messy.fillna(messy["Score"].mean()))   # fill with the mean
print(messy.dropna())              # drop rows with any missing value
```
