```python
import numpy as np

students = np.array(["Tanishq", "Rahul", "Meena", "Kabir", "Isha"])
subjects = np.array(["Maths", "Science", "English", "Computer"])

marks = np.array([
    [88, 92, 79, 95],
    [70, 65, 80, 72],
    [91, 85, 89, 94],
    [55, 60, 58, 62],
    [78, 82, 90, 85],
])

print("Shape:", marks.shape)
print("Asha's marks:", marks[0])
print("Maths column:", marks[:, 0])
print("Last two students, first two subjects:\n", marks[-2:, :2])


print("Maths > 85:", students[marks[:, 0] > 85])

student_avg = marks.mean(axis=1)      # one value per student
subject_avg = marks.mean(axis=0)      # one value per subject
for name, avg in zip(students, student_avg):
    print(f"{name:<8}{avg:.1f}")
print("Topper:", students[student_avg.argmax()])
print("Easiest subject:", subjects[subject_avg.argmax()])

# Broadcasting: add grace marks per subject, capped at 100
grace = np.array([2, 0, 3, 1])
final = np.minimum(marks + grace, 100)
print(final)

# Broadcasting: z-score of every subject (column)
z = (marks - marks.mean(axis=0)) / marks.std(axis=0)
print(z.round(2))

# Students below 60 in any subject
needs_help = students[(marks < 60).any(axis=1)]
print("Needs support:", needs_help)
```
