```python
import pandas as pd
import numpy as np
```

```python
df=pd.DataFrame({
    "name" : ["Ayush", "Sakshi", "Abhisekh", "Ankit", "Rahul"],
    "age" : [20, 21, 19, 18, 22],
    "gender" : ["M", "F", "M", "M", "M"],
    "city" : ["Delhi", "Mumbai", "Bangalore", "Chennai", "Kolkata"],
    "math":    [88, 72, 95, 60, 81],
    "reading": [91, 70, 93, 65, 85],
    "prep":    ["yes", "no", "yes", "no", "yes"],

})
print(df)
```

```python
df["math"]                  
df[["name", "math"]]        
df.head(3)                  
df.loc[2, "name"]           
df.iloc[0:3, 0:1]            
```

```python
df[(df["city"] == "Delhi") & (df["math"] >= 60)]     
df[(df["math"] < 55) | (df["math"] > 90)]             
df[~(df["prep"] == "yes")]                            
```

```python
df[df["city"].isin(["Pune", "Delhi"])]      
df[df["math"].between(70, 90)]              
df[df["name"].str.startswith("S")]          
df[df["name"].str.contains("a")]            
missing = df[df["math"].isna()]             
missing_free = df[df["math"].notna()] 
print(missing)

```

```python
df.query("math > 80 and reading > 85")
df.query("city == 'Delhi' and prep == 'no'")
limit = 75
df.query("math > @limit")                    # @ lets you use a Python variable
```

```python
df.loc[df["math"] > 80, ["name", "city", "math"]]
```

```python
df.nlargest(3, "math")                
df.nsmallest(2, "reading")            
df.sort_values("math", ascending=True).head(3)
```

```python
df2 = df.copy()                                        
df2.loc[df2["math"] < 50, "math"] = 50                 
df2["result"] = np.where(df2["math"] >= 60, "pass", "fail")
df2["band"] = pd.cut(df2["math"], [0, 59, 79, 100], labels=["low", "mid", "high"])
print(df2)
print(df2["band"].value_counts())
```

```python
#group
df.groupby("city")["math"].mean()
```

```python
df.groupby("city").size()             
df.groupby("city")["math"].count()    
df["city"].value_counts()             
df["prep"].value_counts(normalize=True) 
print(df)
print(df.groupby("city")["math"].mean())
```

```python
df.groupby("city")["math"].agg(["mean", "min", "max"])

summary = df.groupby("city").agg(
    students=("name", "count"),
    avg_math=("math", "mean"),
    avg_reading=("reading", "mean"),
    best_math=("math", "max"),
).round(2)
print(summary)
```

```python
df.groupby(["city", "prep"])["math"].mean()                
df.groupby(["city", "prep"])["math"].mean().unstack()      
```

```python
df["city_avg_math"] = df.groupby("city")["math"].transform("mean")
df["vs_city"] = (df["math"] - df["city_avg_math"]).round(1)
print(df)
```

```python
# Top 2 in each city
df.sort_values("math", ascending=False).groupby("city").head(2)

df.loc[df.groupby("city")["math"].idxmax(), ["city", "name", "math"]]

df.groupby("city").filter(lambda g: len(g) >= 4)         #
```

```python
df.pivot_table(values="math", index="city", columns="prep", aggfunc="mean", margins=True)
pd.crosstab(df["city"], df["prep"])
```
