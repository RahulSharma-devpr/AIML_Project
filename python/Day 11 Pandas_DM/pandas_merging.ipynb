```python
import pandas as pd 
mentors = pd.DataFrame({
    "city":   ["Delhi", "Mumbai", "Chennai"],
    "mentor": ["Mr. Rao", "Ms. Iyer", "Mr. Das"],
    "Course": ["Data Science", "Web Development", "Data Analytics"]
})
print(mentors)
```

```python
df = pd.DataFrame({
    "city": ["Delhi", "Mumbai", "Kolkata", "Chennai"],
    "mentors": ["Rahul", "Ramesh", "Suresh", "Mahesh"],
    "Course": ["Data Science", "Web Development", "Data Analytics", "Data Science"]})
pd.merge(df, mentors, on="city", how="inner")
pd.merge(df, mentors, on="city", how="left")
pd.merge(df, mentors, on="city", how="right")
pd.merge(df, mentors, on="city", how="outer")
```

```python
merged = pd.merge(df, mentors, on="city", how="left",indicator=True, validate="many_to_one")
print(merged)
print(len(df), len(merged))                         
print(merged["_merge"].value_counts())             
print(merged[merged["_merge"] == "left_only"])     
```
