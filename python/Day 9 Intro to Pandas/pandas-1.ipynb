```python
marks = pd.Series([88, 72, 95, 60],
                  index=["Yash", "Rahul", "kshitiz", "Kabir"],
                  name="Maths")
print(marks)

print(marks["Rahul"])          
print(marks.iloc[2])           
print(marks[["Rahul", "Kabir"]])
print(marks[marks > 70])       

print(marks.mean(), marks.max(), marks.idxmax())   
print(marks + 5)               
print(marks.sort_values(ascending=False))
```

```python
s1 = pd.Series({"a": 1, "b": 2, "c": 3})     # from a dictionary
s2 = pd.Series(np.arange(5))                  # from a NumPy array, default index 0..4
s3 = pd.Series([1, np.nan, 3])                # np.nan marks a missing value

print(s3.isna())            # [False  True False]
print(s3.fillna(0))         # [1. 0. 3.]
```

```python

import pandas as pd
data = {
    "Name":  ["Asha", "Rahul", "Meena", "Kabir", "Isha"],
    "Age":   [21, 22, 20, 23, 21],
    "City":  ["Delhi", "Mumbai", "Delhi", "Pune", "Mumbai"],
    "Marks": [88, 72, 95, 60, 81],
}
df = pd.DataFrame(data)

print(df["Name"])                  # one column -> Series
print(df[["Name", "Marks"]])       # several columns -> DataFrame (note the double brackets)

# loc: by label (end is INCLUDED)
print(df.loc[0])                       # row with index label 0
print(df.loc[0, "Name"])               # Asha
print(df.loc[1:3, ["Name", "Marks"]])  # rows 1, 2 and 3

# iloc: by position (end is EXCLUDED, like Python slicing)
print(df.iloc[0:2, 0:2])               # first 2 rows, first 2 columns
print(df.iloc[-1])                     # last row
```
