```python
import numpy as np

size = np.array([500, 750, 1000, 1250, 1500, 1800, 2000])    # sq ft
price = np.array([25, 37, 50, 61, 76, 90, 99])               # lakhs


X = np.column_stack((np.ones(len(size)), size))


theta = np.linalg.inv(X.T @ X) @ (X.T @ price)
intercept, slope = theta
print(f"Price = {intercept:.2f} + {slope:.4f} x size")


predicted = X @ theta
mse = np.mean((price - predicted) ** 2)
ss_res = np.sum((price - predicted) ** 2)
ss_tot = np.sum((price - price.mean()) ** 2)
r2 = 1 - ss_res / ss_tot
print(f"MSE: {mse:.2f}   R squared: {r2:.4f}")


new_size = 1600
print(f"Predicted price for {new_size} sq ft: {intercept + slope * new_size:.1f} lakhs")   # roughly 80


print(np.polyfit(size, price, 1))      # [slope, intercept]
```
