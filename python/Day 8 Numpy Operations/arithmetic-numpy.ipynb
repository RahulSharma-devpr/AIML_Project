```python
#Element-wise arithmetic and comparison
import numpy as np
a = np.array([1,2,3,4])
b = np.array([21,22,23,24])

print(a+b)
print(a*b)
print(a/b)
print(a-b)
print(a ** 2)    
print(a > 2)      
```

```python
#Universal Function
print(np.sqrt(a))         
print(np.exp(a))           
print(np.log(b))          
print(np.abs(np.array([-3, 4, -5])))     
print(np.round(3.14159, 2), np.floor(2.7), np.ceil(2.1))   
```

```python
#Aggregations and the axis argument
import numpy as np
data = np.array([[4, 8, 6],
                 [2, 9, 5]])

print(data.sum(), data.mean(), data.max())    
print(data.sum(axis=0))       
print(data.sum(axis=1))       
print(data.argmax())          
print(np.median(data), np.std(data), np.var(data))
print(np.cumsum([1, 2, 3, 4]))   
```

```python
#Conditions with where, and counting
scores = np.array([45, 82, 67, 91, 38])

print(np.where(scores >= 50, "Pass", "Fail"))   
print(np.sum(scores >= 50))                      
print(scores[(scores > 40) & (scores < 90)])  
print(np.clip(scores, 40, 90))                   
```

```python
#Sorting and unique values
x = np.array([3, 1, 2, 3, 1])

print(np.sort(x))                         
print(np.argsort(x))                      
values, counts = np.unique(x, return_counts=True)
print(values, counts)                     
```

```python
#Matrix operations
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print(A * B)              
print(A @ B)             
print(A.T)                
print(np.linalg.inv(A))   
print(np.linalg.det(A))   


coeffs = np.array([[2, 1], [1, 3]])
constants = np.array([5, 10])
print(np.linalg.solve(coeffs, constants))  
```

```python
#time comparison between list and numpy array
import time

nums = list(range(1_000_000))
arr = np.arange(1_000_000)

start = time.time()
[n * 2 for n in nums]
print("List :", time.time() - start)

start = time.time()
arr * 2
print("NumPy:", time.time() - start)      
```

```python
# Min-max scaling (squeeze values into 0 to 1)
x = np.array([20, 35, 50, 80, 100])
scaled = (x - x.min()) / (x.max() - x.min())
print(scaled)          # [0.     0.1875 0.375  0.75   1.    ]

# Z-score standardisation (mean 0, std 1)
z = (x - x.mean()) / x.std()

# Euclidean distance between two points
p1 = np.array([1, 2])
p2 = np.array([4, 6])
print(np.sqrt(np.sum((p1 - p2) ** 2)))    # 5.0

# 3-day moving average
prices = np.array([10, 12, 14, 13, 15, 17])
print(np.convolve(prices, np.ones(3) / 3, mode="valid"))   # [12. 13. 14. 15.]

# Percentage of values above the average
values = np.array([5, 9, 3, 8, 6])
print((values > values.mean()).mean() * 100)    # 40.0
```
